```python
import warnings
import gymnasium as gym
```

```python
gym.register(
    id="CoSwap", entry_point="utils.co_routing_env:CoRoutingEnv")
env = gym.make(
    id="CoSwap", start_level=4)
warnings.filterwarnings("ignore", category=UserWarning)
```

```python
state, info = env.reset()
print(state)
print(info)
screen = env.render()
screen
```

```python
state, reward, terminated, truncated, info = env.step((2, 3))
print(reward)
print(state)
print(terminated, truncated)
```
