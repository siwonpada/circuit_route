```python
from stable_baselines3 import PPO
from sb3_contrib.ppo_mask import MaskablePPO

import gymnasium as gym
```

```python
gym.register(id="SABRE-co", entry_point="RL_SB3_optuna_dense_co_edge:CoRoutingEnv")
env = gym.make("SABRE-co", start_level=4)
```

```python
model = MaskablePPO.load("./optuna_sb3_dense_co_edge/saves/rl_model_0.zip")

state, info = env.reset()
done = False
while not done:
    action, _ = model.predict(state)
    state, reward, terminated, truncated, info = env.step(action)
    done = terminated or truncated
env.render()
```
