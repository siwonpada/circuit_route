```python
from stable_baselines3 import PPO

import gymnasium as gym
```

```python
gym.register(id="SABRE-wogt", entry_point="utils.sabre_env_wogt_wc:SabreSwapEnv")
env = gym.make("SABRE-wogt", start_level=10)
```

```python
model = PPO.load("./optuna_sb3_wogt/saves/rl_model_11.zip")

state, info = env.reset()
done = False
while not done:
    action, _ = model.predict(state)
    state, reward, terminated, truncated, info = env.step(action)
    done = terminated or truncated
env.render()
```
